# **Gráficos**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["volume"], bins=30)
plt.title("Distribuição do Volume")
plt.xlabel("Volume")
plt.ylabel("Quantidade")
plt.show()
```

O histograma evidencia que a maior parte dos mercados possui baixo volume, enquanto poucos concentram grande parte das negociações, caracterizando uma distribuição assimétrica.

```python
df["closed"].value_counts().plot(kind="bar")

plt.title("Mercados Encerrados")
plt.show()
```

```python
top10 = df.sort_values("volume", ascending=False).head(10)

plt.figure(figsize=(12,5))
plt.bar(top10["question"], top10["volume"])
plt.xticks(rotation=90)
plt.show()
```
