```python
# criando arrays numpy a partir de algumas colunas numéricas do DataFrame
# substituindo valores nulos por 0 por questões de consistência de cálculo!

# preço unitário
preco_unitario = dados['preco_unitario'].fillna(0).to_numpy()
print("Preço unitário:", preco_unitario)

# custo unitário
custo_unitario = dados['custo_unitario'].fillna(0).to_numpy()
print("Custo unitário:", custo_unitario)

# quantidade
quantidade = dados['quantidade'].fillna(0).to_numpy()
print("Quantidade:", quantidade)

# desconto
desconto = dados['desconto_pct'].fillna(0).to_numpy()
print("Desconto (%):", desconto)

# receita total
receita_total = dados['receita_total'].fillna(0).to_numpy()
print("Receita total:", receita_total)

# frete
frete = dados['frete'].fillna(0).to_numpy()
print("Frete:", frete)
```

### B. Cálculos com operações vetorizadas

```python
#1. valor bruto da venda
valor_bruto_venda = preco_unitario * quantidade
print("Valor bruto da venda:", valor_bruto_venda)

#2. valor do desconto
valor_desconto = valor_bruto_venda * desconto / 100
print("Valor do desconto:", valor_desconto)

#3. custo total
custo_total = custo_unitario * quantidade
print("Custo total:", custo_total)

#4. lucro estimado
lucro_estimado = valor_bruto_venda - (valor_desconto + custo_total + frete)
print("Lucro estimado:", lucro_estimado)
```

### C. Máscaras booleanas


```python
#1. vendas com receita acima da média
mascara_receita_acima_media = receita_total > np.mean(receita_total)

#2. vendas com prejuízo
mascara_prejuizo = lucro_estimado < 0

#3. vendas com desconto alto
mascara_desconto_alto = desconto > 20
```

```python
print(f'Quantidade de vendas com receita acima da média: {len(dados[mascara_receita_acima_media])} ({round((len(dados[mascara_receita_acima_media]['receita_total']) / len(dados)) * 100, 2)} %)')

print(f'Quantidade de vendas com desconto alto: {sum(mascara_desconto_alto)} ({round((sum(mascara_desconto_alto) / len(dados)) * 100, 2)} %)')
```

### D. Agregações

```python
#1. receita média
receita_media = np.mean(receita_total)
print("Receita média: R$", round(receita_media, 2))

#2. maior receita
maior_receita = np.max(receita_total)
print("Maior receita: R$", round(maior_receita, 2))

#3. menor receita
menor_receita = np.min(receita_total)
print("Menor receita: R$", round(menor_receita, 2))

#4. lucro médio
lucro_media = np.mean(lucro_estimado)
print("Lucro médio: R$", round(lucro_media, 2))

#5. quantidade de vendas com prejuízo
quantidade_prejuizo = sum(mascara_prejuizo)
print(f'Quantidade de vendas com prejuízo: {sum(mascara_prejuizo)} itens ({round((sum(mascara_prejuizo) / len(dados)) * 100, 2) } %)')
```
