1)

```python
vendas_mes = df.groupby('mes')['receita_total'].sum().reset_index()

vendas_mes_ano = (
    df.groupby(['ano', 'mes'])['receita_total']
      .sum()
      .reset_index()
)

# Criação dos gráficos
fig, ax = plt.subplots(2, 1, figsize=(14,8))

# Gráfico 1 - Sazonalidade geral
sns.lineplot(
    data=vendas_mes,
    x='mes',
    y='receita_total',
    marker='o',
    ax=ax[0]
)

ax[0].set_title('Vendas por Mês (Sazonalidade)')
ax[0].set_xlabel('Mês')
ax[0].set_ylabel('Vendas')

# Gráfico 2 - Cada ano separado
sns.lineplot(
    data=vendas_mes_ano,
    x='mes',
    y='receita_total',
    hue='ano',
    marker='o',
    ax=ax[1]
)

ax[1].set_title('Vendas por Mês em Cada Ano')
ax[1].set_xlabel('Mês')
ax[1].set_ylabel('Vendas')

plt.tight_layout()
plt.show()
```

Vemos alguns períodos de sazonalidade importantes para a empresa: Em fevereiro, tanto em 2022 ou em 2023, temos um volume de vendas muito abaixo em comparação com janeiro, o que nos indica um padrão muito importante para esse começo de ano. Além disso, temos uma uniformidade muito boa dos padrões, tanto o ano de 2022 e 2023 seguem quase na mesma proporção, o que é um ponto positivo para prever novas demandas e necessidades.

2)

```python
QTD_mes = df.groupby('mes')['quantidade'].sum().reset_index()

QTD_mes_ano = (
    df.groupby(['ano', 'mes'])['quantidade']
      .sum()
      .reset_index()
)

# Criação dos gráficos
fig, ax = plt.subplots(2, 1, figsize=(14,8))

# Gráfico 1 - Sazonalidade geral
sns.lineplot(
    data=QTD_mes,
    x='mes',
    y='quantidade',
    marker='o',
    ax=ax[0]
)

ax[0].set_title('QTD por Mês (Sazonalidade)')
ax[0].set_xlabel('Mês')
ax[0].set_ylabel('QTD')

# Gráfico 2 - Cada ano separado
sns.lineplot(
    data=QTD_mes_ano,
    x='mes',
    y='quantidade',
    hue='ano',
    marker='o',
    ax=ax[1]
)

ax[1].set_title('QTD por Mês em Cada Ano')
ax[1].set_xlabel('Mês')
ax[1].set_ylabel('QTD')

plt.tight_layout()
plt.show()
```

Parecido com o gráfico 1, vemos um padrão muito parecido em ambos os anos, o que é um ponto positivio para pensar em novas práticas de mercado. Além disso, vemos que no 2° e 3° trimestre temos um volume muito maior de quantidade vendida, o que nos pode indicar que deve ser de produtos com  ticket médio menor, visto que o faturamento total do 3° mês até o 10° mês segue quase o mesmo.

3)

```python
Lucro_mes = df.groupby('mes')['lucro'].sum().reset_index()

Lucro_mes_ano = (
    df.groupby(['ano', 'mes'])['lucro']
      .sum()
      .reset_index()
)

# Criação dos gráficos
fig, ax = plt.subplots(2, 1, figsize=(14,8))

# Gráfico 1 - Sazonalidade geral
sns.lineplot(
    data=Lucro_mes,
    x='mes',
    y='lucro',
    marker='o',
    ax=ax[0]
)

ax[0].set_title('Lucro por Mês (Sazonalidade)')
ax[0].set_xlabel('Mês')
ax[0].set_ylabel('Lucro')

# Gráfico 2 - Cada ano separado
sns.lineplot(
    data=Lucro_mes_ano,
    x='mes',
    y='lucro',
    hue='ano',
    marker='o',
    ax=ax[1]
)

ax[1].set_title('Lucro por Mês em Cada Ano')
ax[1].set_xlabel('Mês')
ax[1].set_ylabel('Lucro')

plt.tight_layout()
plt.show()
```

Apesar de termos o maior volume faturado em janeiro e dezembro, nosso maior lucro veio no mês de setembro, o que nos pode indicar que há um mix de produtos nesse mês que tem uma margem de lucro maior que os demais. Além de ter um padrão peculiar, vendo que em todo fechamento de trimestre (3, 6, 9 e 12) temos um crescimento bem grande do lucro em 2023.
