## Entregável 1 NumPy

##### A. Criando arrays

```python
preco_unit = np.array(df['preco_unitario'])
custo_unit = np.array(df['custo_unitario'])
margem_pct = np.array(df['margem_pct'])
quantidade = np.array(df['quantidade'])
desconto_pct = np.array(df['desconto_pct'])
receita_total = np.array(df['receita_total'])
frete = np.array(df['frete'])
avaliacao_cliente= np.array(df['avaliacao_cliente'])
```

##### B. Cálculos com operações vetorizadas

```python
'''1. valor bruto da venda
2. valor do desconto
3. custo total
4. lucro estimado'''

valor_bruto_venda = preco_unit * quantidade
valor_desconto = valor_bruto_venda * desconto_pct/100
custo_total = custo_unit * quantidade
lucro_estimado = receita_total - custo_total 
markup = preco_unit / custo_unit

valor_bruto_venda, valor_desconto, custo_total, lucro_estimado, markup
```

##### C. Máscaras booleanas

```python
np.nanmean(receita_total)
vendas_acima_media = receita_total > np.nanmean(receita_total)
prejuizo = lucro_estimado < 0 
desconto_alto = desconto_pct > 0.3
```

```python
vendas_acima_media = receita_total > np.nanmean(receita_total)
print(vendas_acima_media)
print(receita_total[vendas_acima_media])
print(np.nanmean(receita_total[vendas_acima_media]))
```

```python
print(np.nanmean(lucro_estimado))
print(lucro_estimado[prejuizo])
print(np.nanmean(lucro_estimado[prejuizo]))
```

##### D. Agregações

```python
print(f'a média da receita foi: {round(np.nanmean(receita_total), 2)}')
print(f'a maior receita foi: {receita_total.max()}')
print(f'a menor receita foi: {receita_total.min()}')
print(f'o lucro médio foi: {round(np.nanmean(lucro_estimado), 2)}')
print(f'o maior lucro foi: {round(lucro_estimado.max(), 2)}')
print(f'a média da quantidade de produtos vendidos foi: {round(np.nanmean(quantidade), 2)}')
```

## Entregável 3 Seaborn

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)

sns.set_theme(context="notebook", palette="deep")
```

```python
sns.barplot(data=df,x='regiao',y='receita_total')
```

Podemos ver que a região que mais gerou receita foi a região norte.

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df,x='cidade',y='receita_total')
plt.ylabel('Receita Total')
plt.xlabel('')
plt.show()
```

Podemos ver que a cidade que mais gerou receita foi Recife.

```python
plt.figure(figsize=(7, 6))
sns.barplot(data=df,x='regiao',y='lucro_estimado',hue='faixa_de_lucro')
plt.ylabel('Lucro Estimado')
plt.xlabel('')
plt.show()
```

As faixas de lucro entre as regiões são parecidas, mas com o centro-oeste possuindo uma concentração maior de faixa de lucro alta.

##### B. Gráfico de linha

```python
plt.figure(figsize=(12, 5))
df['mes'] = pd.Categorical(df['mes'], ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro'])
sns.lineplot(data=df,x='mes',y='receita_total',linewidth=2.5)
plt.xlabel('')
plt.ylabel('Receita Total')
plt.show()
```

Podemos ver uma concentração de receita no trimestre Novembro-Dezembro-Janeiro, provavelmente devido ao Natal, pagamento do 13º, entre outros motivos.

##### C. Gráfico de dispersão

```python
amostra = df.sample(frac=0.3, random_state=110)
sns.scatterplot(data=amostra,x='quantidade',y='receita_total',hue='categoria')
plt.xlabel('Quantidade')
plt.ylabel('Receita Total')
plt.show()
```

Desconsiderando os outliers, podemos ver que os eletrônicos são os que compõem maior receita total mesmo em pouca quantidade, refletindo o quão caros são em relações à produtos de outras categorias.

##### D. Mapa de calor

```python
filtrado_df = df[['quantidade', 'receita_total', 'frete', 'avaliacao_cliente', 'lucro_estimado']]
plt.figure(figsize = (10,8))
sns.heatmap(filtrado_df.corr(), cmap = 'coolwarm')
plt.show()
```

Podemos ver uma considerável correlação positiva entre receita total e lucro estimado. Há também uma leve correlação entre quantidade com receita total e lucro estimado, já que quanto mais produtos comprados, maior a receita/lucro.
