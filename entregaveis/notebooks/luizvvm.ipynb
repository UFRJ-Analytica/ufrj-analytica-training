### A. Convertendo algumas colunas pra Numpy para realizar a primeira atividade

```python
preco_unitario = np.array(vendas['preco_unitario'])
custo_unitario = np.array(vendas['custo_unitario'])
quantidade = np.array(vendas['quantidade'])
desconto = np.array(vendas['desconto_pct'])
receita_total = np.array(vendas['receita_total'])
frete = np.array(vendas['frete'])
```

### B.

```python
# valor bruto da venda
valor_bruto = [preco_unitario * quantidade]
    
# valor do desconto
desconto_preco = (valor_bruto * (desconto/100))

#custo total
custo_total = (custo_unitario * quantidade + frete)

#lucro estimado
lucro_estimado = (valor_bruto - desconto - custo_total)
```

### C.

```python
# Vendas com receita acima da média
receita_media = np.mean(receita_total)
vendas_receita_acima_media = receita_total[receita_total > receita_media]

# vendas com prejuizo
prejuizos = lucro_estimado[lucro_estimado < 0]

# vendas com desconto alto
desconto_medio = np.mean(desconto_preco)
descontos_altos = desconto_preco[desconto_preco > desconto_medio]

#visualizar
print("________Vendas com receita acima da média________")
for x in vendas_receita_acima_media:
    print(x)
    
print("________vendas com prejuizo________")
for x in prejuizos:
    print(x)
    
print("________vendas com desconto alto________")
for x in descontos_altos:
    print(x)
```

### D

```python
# Receita Media (já calculei ali em cima)
receita_media

# Maior receita
maior_receita = np.max(receita_total)

# Menor receita
menor_receita = np.min(receita_total)

# Lucro Médio
lucro_medio = np.mean(lucro_estimado)

# quantidade de vendas com prejuizo
qntd_prejuizos = np.count_nonzero(prejuizos)
```

### B

```python
vendas['data_venda'] = pd.to_datetime(vendas['data_venda'], format='mixed')

vendas['canal_venda'].unique()
vendas['canal_venda'] = vendas['canal_venda'].str.lower()

vendas['avaliacao_cliente'] = vendas['avaliacao_cliente'].fillna('Não informado')
```

### C.

```python
vendas['ano_da_venda'] = vendas['data_venda'].dt.year
vendas['mes_da_venda'] = vendas['data_venda'].dt.month
vendas['lucro_total'] = (vendas['preco_unitario'] - (vendas['preco_unitario'] * (vendas['desconto_pct']/100)) - vendas['custo_unitario']) * vendas['quantidade']
vendas['receita_liquida'] = vendas['receita_total'] - vendas['preco_unitario'] * (vendas['desconto_pct']/100) * vendas['quantidade']

def faixa_receita(x):
    if(x > np.mean(x)):
        return True
    else:
        return False

vendas['faixa_de_receita'] = vendas['receita_liquida'].apply(faixa_receita)
```

# 3. Seaborn

### A.

```python
receita_categoria = vendas.groupby('categoria')['receita_total'].sum().sort_values(ascending=False).reset_index()
receita_categoria_grafico = sns.barplot(
    data=receita_categoria,
    x='categoria',
    y='receita_total',
    palette='Set1'
)
plt.title('Receita por categoria')
plt.xlabel("Categoria")
plt.ylabel('Receita')
sns.despine()

plt.show()
```

```python
lucro_mensal = vendas.groupby('mes')['lucro_total'].sum().reset_index()
sns.lineplot(
    data=lucro_mensal,
    x='mes',
    y='lucro_total',
    color='blue',
    marker='o'
)
plt.title('Lucro mensal')
plt.xlabel('Mês')
plt.ylabel('Lucro total')
sns.despine()

plt.show()
```

```python
sns.scatterplot(
    data=vendas,
    x='preco_unitario',
    y='quantidade',
    color='blue'
)

plt.title('Preço unitário por Quantidade')
plt.xlabel('Preço unitário')
plt.ylabel('Quantidade')

sns.despine()

plt.show()
```

```python
sns.heatmap(
    vendas[[
            'preco_unitario',
            'custo_unitario',
            'quantidade',
            'desconto_pct',
            'receita_total',
            'lucro_total',
            'frete'
        ]].corr(),
    annot=True,
    cmap='Greens',
    fmt='.2f'
)

plt.title('Correlação entre as variáveis')

plt.show()
```
