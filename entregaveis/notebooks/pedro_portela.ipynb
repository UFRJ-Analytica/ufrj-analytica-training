# Atividade de vendas no Brasil

## Entregável 1 NumPy

### A. Criando arrays

```python
#Lendo o arquivo csv com pandas
df = pd.read_csv("vendas_brasil_1.csv")

#Criando arrays para as colunas numéricas do dataset usando numpy
preco_unitario_np = df['preco_unitario'].to_numpy()

custo_unitario_np = df['custo_unitario'].to_numpy()

quantidade_np = df['quantidade'].to_numpy()

desconto_pct_np = df['desconto_pct'].to_numpy()

receita_total_np = df['receita_total'].to_numpy()

frete_np = df['frete'].to_numpy()
print("Dataset")
print(df)
print(f'Os arrays criados a partir deste dataset foram:\nPreço unitario:{preco_unitario_np}\ncusto unitario:{custo_unitario_np}\ndesconto pct:{desconto_pct_np}\nquantidade:{quantidade_np}\nreceita total:{receita_total_np}\nfrete:{frete_np} ')
```

### B. Cálculos com operações vetorizadas

```python
#calculando valores importantes sobre o dataset
valor_bruto = preco_unitario_np * quantidade_np
valor_desconto = valor_bruto * (desconto_pct_np/100)
custo_total = custo_unitario_np * quantidade_np
lucro_estimado = receita_total_np - custo_total

print(f'valor bruto dos produtos: {valor_bruto}')
print(f'desconto dos produtos: {valor_desconto}')
print(f'custo total da produção dos produtos: {custo_total}')
print(f'lucro estimado da venda de cada produto: {lucro_estimado}')
```

### C. Máscaras booleanas

```python
#Filtrando situações inusitadas do dataset

media_receita = np.mean(receita_total_np)
vendas_acima_media = receita_total_np[receita_total_np > media_receita]

vendas_prejuizo = lucro_estimado[lucro_estimado < 0]

media_desconto = np.mean(desconto_pct_np)
vendas_desconto_alta = desconto_pct_np[desconto_pct_np > media_desconto]

print(f'quantidade de vendas com receita acima da média: {len(vendas_acima_media)}')
print(f'quantidade vendas que geraram prejuízo: {len(vendas_prejuizo)}')
print(f'quantidade de vendas com desconto acima da média: {len(vendas_desconto_alta)}')
print(f'\nquantidade de vendas totais: {len(df['data_venda'])}')
```

### D. Agregações

```python
#calculando medidas gerais para ánalise

media_das_receitas = np.mean(receita_total_np)

maior_receita = np.max(receita_total_np)

menor_receita = np.min(receita_total_np)

media_dos_lucros = np.mean(lucro_estimado)

quantidade_vendas_prejuizo = len(vendas_prejuizo)

print(f'media das receitas: {media_das_receitas}')
print(f'maior receita: {maior_receita}')
print(f'menor receita: {menor_receita}')
print(f'media dos lucros: {media_dos_lucros}')
print(f'quantidade de vendas com prejuizo: {quantidade_vendas_prejuizo}')
```

### Conclusões gerais


Observando o dataset apenas com funções matematicas e de array do numpy podemos observar algumas inconsistências.

Dentro das vendas que análisamos vemos que temos alguns outliers tendo em vista que a média das receitas é 1891 enquanto a maior e menor receita que vimos são respectivamente 58430 e 3.

Podemos notar tambem que existem 2168 produtos onde o lucro obtido é negativo, gerando necessidade de uma maior atenção e ánalise em tais objetos

### B. Limpeza e tratamento

```python
#convertendo a coluna data para datatype
df['data_venda'] = pd.to_datetime(df['data_venda'], format='mixed')

#padronizando os nomes das cidades
df['cidade'] = df['cidade'].str.upper()
mapeamento_cidades = {
    'RIO': 'RIO DE JANEIRO',
    'RJ': 'RIO DE JANEIRO',
    'RIO DEJANEIRO': 'RIO DE JANEIRO',
    'CURTIBA': 'CURITIBA',
    'SP': 'SÃO PAULO',
    'S. PAULO': 'SÃO PAULO',
    'SAO PAULO': 'SÃO PAULO',
    'BH': 'BELO HORIZONTE',
    'B. HORIZONTE': 'BELO HORIZONTE',
    'SLAVADOR': 'SALVADOR'
    }
df['cidade'] = df['cidade'].str.strip()
df['cidade'] = df['cidade'].replace(mapeamento_cidades)
df['cidade'].value_counts()

#Analisando a coluna avaliacao_cliente para decidir a melhor forma de substituir dados faltantes
"""
print(f'Analisando a coluna "avaliacao_cliente"')
print(f'média: {df['avaliacao_cliente'].mean()}')
print(f'maior coluna: {df['avaliacao_cliente'].max()}')
print(f'menor coluna: {df['avaliacao_cliente'].min()}')
"""
media_avaliacao_clientes = df['avaliacao_cliente'].mean()
df['avaliacao_cliente'] = df['avaliacao_cliente'].fillna(media_avaliacao_clientes)
df.isna().sum()

#Analisando a coluna frete para decidir a melhor forma de substituir dados faltantes
'''
print(f'Analisando a coluna "frete"')
print(f'{df['frete']}')
'''
df['frete'] = df['frete'].fillna(0)
df.isna().sum()
df['data_venda']
```

Decidi substituir os dados faltantes da coluna 'frete' por 0 pois acho plausível supor que os dados nulos que aparecem são devidos a não necessidade de entrega do produto fazendo o frete não existir.

Como a coluna 'avaliacao_clientes' não possuia outliers grandes a melhor opção para substituir os dados faltantes é muda-los pela média da própria coluna.

Dessa maneira evitamos um viés muito grande na ánalise do noss0 dataset

### C. Criação de novas colunas

```python
#Criando uma coluna para o ano das vendas
df['ano_venda'] = df['data_venda'].dt.year
df[['ano_venda','data_venda']]

#Criando uma coluna para o lucro total
df['lucro_total'] = lucro_estimado #Variavel "lucro_estimado" criada em 1.B Calculos com operações vetorizadas
df['lucro_total']

#Criando uma coluna para receita liquida
df['receita_liquida'] = df['receita_total'] - df['desconto_pct'] - df['frete']
df[['receita_liquida', 'receita_total']].head()

#Criando uma coluna para faixa de receita
df['faixa_receita'] = df['receita_total'].apply(lambda x: 'alta' if x > media_das_receitas else 'baixa')
df[['faixa_receita', 'receita_total']]
```

### D. Filtros

```python
#Criando um filtro para saber quais receitas foram altas
receita_alta = df[df['faixa_receita'] == 'alta']
receita_alta

#Filtro para saber quais vendas tiveram devolução
vendas_com_devolucao = df[df['devolucao'] == True]
df['devolucao'].value_counts() #Nenhum dado fora do comum
vendas_com_devolucao

#Filtrando quais vendas foram da categoria Alimentação
vendas_alimentacao = df[df['categoria'] == 'Alimentação']
vendas_alimentacao

print(f'Os filtros criados foram:\nReceitas altas:\n{receita_alta}\nitens devolvidos\n{vendas_com_devolucao}\nquais vendas foram para Alimentação:\n{vendas_alimentacao}')
```

Com os filtros pudemos encontrar quais foram as receitas que tiveram um alto valor(valor acima da média), temos uma noção de quais vendas tiveram devolução e quantas e quais vendas foram da categoria de Alimetnação

### E. Agrupamentos com groupby

```python
#Calculando a receita que cada categoria gerou
receita_categoria = df.groupby('categoria')['receita_total'].sum()
receita_categoria

#Descobrindo qual canal de venda vendeu em maior quantidade
canal_mais_vendas = df.groupby('canal_venda')['quantidade'].sum()
canal_mais_vendas

#Analisando qual cidade teve maior receita
cidade_maior_receita = df.groupby('cidade')['receita_total'].sum()
cidade_maior_receita
```

### F. Merge

```python
#Criando uma tabela de regiões para ligar cada cidade do dataset a uma região
regioes = {
    'cidade': ['RIO DE JANEIRO', 'SÃO PAULO', 'BELO HORIZONTE', 'CURITIBA', 'FORTALEZA', 'GOIÂNIA', 'MANAUS', 'PORTO ALEGRE','RECIFE', 'SALVADOR'],
    'regioes': ['SUDESTE', 'SUDESTE', 'SUDESTE', 'SUL', 'NORDESTE', 'CENTRO-OESTE', 'NORTE', 'SUL', 'NORDESTE', 'NORDESTE']
}
df_regioes = pd.DataFrame(regioes)
df_regioes

df_atualizado = pd.merge(df, df_regioes, on='cidade', how='left')
receita_por_regiao = df_atualizado.groupby('regioes')['receita_total'].sum()
receita_por_regiao
```
